3. Less Memory

```python
lista = range(100)
lista
```

```python
arr11 = np.arange(100)
arr11
```

```python
import sys 
```

```python
print(sys.getsizeof(87)*len(lista)) #2800 bytes
```

```python
print(arr11.itemsize*arr11.size) #size se no of item nikalta hai
```

```python
import time
```

```python
x = range(100000)
y = range(100000, 200000)


start_time = time.time()

c = [(x + y) for x,y in zip(x,y) ]

print(time.time() - start_time)
```

```python
a = np.arange(100000)
b = np.arange(100000, 200000)

start_time = time.time()
c = a + b

print(time.time()- start_time)
```

#### Adding 2 zeroes 

```python
x = range(10000000)
y = range(10000000, 20000000)


start_time = time.time()

c = [(x + y) for x,y in zip(x,y) ]

print(time.time() - start_time)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start_time = time.time()
c = a + b

print(time.time()- start_time)
```
