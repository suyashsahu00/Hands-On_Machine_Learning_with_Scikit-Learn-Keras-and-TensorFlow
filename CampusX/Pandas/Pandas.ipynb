## The drop_duplicates() method

```python
# 1. Sabse pehle 'date' column ko datetime object mein convert karein
data['date'] = pd.to_datetime(data['date'])
```

```python
# 2. Data ko date ke chronological order mein sort karein (Ascending)
data_sorted = data.sort_values('date')
```

```python
# 3. Ab drop_duplicates lagayein
# Har season ka final match nikalne ke liye (keep='last'):
final_matches = data_sorted.drop_duplicates(subset=['season'], keep='last')
```

```python
# Har city mein season ka pehla match nikalne ke liye (keep='first'):
first_matches_in_city = data_sorted.drop_duplicates(subset=['city', 'season'], keep='first')
```

```python
# 'date' ko datetime mein badlein aur sort karein
data['date'] = pd.to_datetime(data['date'])
data_sorted = data.sort_values('date')
```

```python
# Har season ka last match (Final Match) nikal kar winners print karein
season_winners = data_sorted.drop_duplicates(subset=['season'], keep='last')[['season', 'winner']]
season_winners.sort_values('season')
```

```python
def run_scored(batsman_name):
    vk = delivery[delivery['batsman'] == batsman_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).index[0]
```

```python
run_scored('DA Warner')
```

```python
run_scored('G Gambhir')
```

```python
run_scored('MS Dhoni')
```
