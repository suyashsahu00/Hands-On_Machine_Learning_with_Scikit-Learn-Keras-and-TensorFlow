```python
import joblib
import numpy as np
import pandas as pd
```

```python
# Load saved models using joblib
ohe_sex = joblib.load('models/ohe_sex.pkl')
ohe_embarked = joblib.load('models/ohe_embarked.pkl')
clf = joblib.load('models/clf.pkl')
```

```python
# Assume user input
# Pclass / gender / age / SibSp / Parch / Fare / Embarked
test_input = pd.DataFrame(
    [[2, 'male', 31.0, 0, 0, 10.5, 'S']],
    columns=['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[['Sex']])
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[['Embarked']])
```

```python
test_input_embarked
```

```python
test_input_age = test_input[['Age']].values
```

```python
test_input_rem = test_input[['Pclass', 'SibSp', 'Parch', 'Fare']].values

test_input_transformed = np.concatenate(
    (test_input_rem, test_input_age, test_input_sex, test_input_embarked),
    axis=1
)
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
